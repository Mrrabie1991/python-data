```python
import numpy as np
```

```python
sensors_data = np.array([[22.1, 23.5, 24.0, 28.7, 21.9, 30.2],
                         [101, 100, 102, 101, 99, 103],
                          [45, 47, 44, 50, 42, 55]])

print(sensors_data.shape)
sensors_data = sensors_data.T
print(sensors_data.shape)
```

```python
print(sensors_data.shape)
print(sensors_data.ndim)
```

```python
print(f"temperature : {sensors_data[:,0]}")
```

```python
print(f"pressure : {sensors_data[:,1]}")
```

```python
temperature_mean = sensors_data[:,0].mean()
print(f"temperature_mean = {temperature_mean:.2f}")
```

```python
max_humidity = sensors_data[:,2].max()
print(f"Max Humidity = {max_humidity}")
```

```python
measurement = sensors_data[sensors_data[:,0] > 25]
print(measurement)
```

```python
offset  = np.array([0.5,-1,2])
calibration_data = sensors_data + offset
print(calibration_data)
```
